```python
clean_df["customer_remarks"] = (
    clean_df["customer_remarks"]
    .astype("string")
    .str.strip()
)
```

response_time_minutes

```python
clean_df["response_time_minutes"] = (
    clean_df["issue_responded"] -
    clean_df["issue_reported_at"]
).dt.total_seconds() / 60
```

```python
clean_df["response_time_minutes"].describe()
```

```python
clean_df[
    clean_df["response_time_minutes"] < 0
].shape
```

response_time_BUCKETS

```python
bins = [-1, 15, 30, 60, 240, 1440, np.inf]

labels = [
    "<15 min",
    "15–30 min",
    "30–60 min",
    "1–4 hours",
    "4–24 hours",
    "24+ hours"
]

clean_df["response_time_bucket"] = pd.cut(
    clean_df["response_time_minutes"],
    bins=bins,
    labels=labels
)
```

```python
def classify_csat(score):
    if pd.isna(score):
        return pd.NA
    elif score <= 2:
        return "Dissatisfied"
    elif score == 3:
        return "Neutral"
    else:
        return "Satisfied"

clean_df["csat_category"] = (
    clean_df["csat_score"].apply(classify_csat)
)
```

```python
clean_df["csat_category"].value_counts()
```
